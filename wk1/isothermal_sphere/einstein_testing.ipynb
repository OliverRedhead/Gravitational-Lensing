```python
from base_classes import DiskSource, Source
from einstein_deflection import EinsteinDeflector
import matplotlib.pyplot as plt
import numpy as np
```

```python
sc = DiskSource(1001, 50)
plane = sc.array
cbar = plt.imshow(plane, origin='lower')
plt.colorbar(cbar)

plt.title("uniform disk source")
plt.xlabel("x")
plt.ylabel("y")

plt.savefig("uniform_disk_source.png")
plt.show()
```

```python
df = EinsteinDeflector(400)
image = df.get_image(source=sc)

cbar = plt.imshow(image, origin='lower')
plt.colorbar(cbar)

plt.title("deflection of a uniform disk about an isothermal, spherical lens")
plt.xlabel("x")
plt.ylabel("y")

plt.savefig('iso_sphere_deflection.png')
plt.show()
```

```python
print(np.sum(sc.array))
print(np.sum(image))
```

now lets try with an image of a galaxy

```python
from PIL import Image

img = Image.open("Milky_Way_illustration-1.jpeg")
arr = np.array(img)

plt.imshow(arr, origin='lower')
plt.show()
```

```python
# arr = np.sum(arr, axis=2)

src = Source(array=arr)
src.pad(1000)
```

```python
plt.imshow(src.array, origin='lower', cmap='inferno')
plt.show()
```

```python
df = EinsteinDeflector(theta_E=1000)

deflected = df.get_image(src)
```

```python
fig, ax = plt.subplots(figsize=(10,10), dpi=120)
ax.imshow(deflected/np.max(deflected), origin='lower')
plt.show()
```

```python
df = EinsteinDeflector(theta_E=1000)
src.array = np.sum(src.array, axis=2)
deflected = df.get_image(src)
```

```python
fig, ax = plt.subplots(figsize=(10,10), dpi=120)
cbar = plt.imshow(deflected/np.max(deflected), origin='lower', cmap='inferno')
plt.colorbar(cbar)
plt.show()
```

```python
from PIL import Image

img = Image.open("Milky_Way_illustration-1.jpeg")
arr = np.array(img)

plt.imshow(arr, origin='lower')
plt.show()
```

what happens when we add a point source?

```python
print(arr[200,200])

arr[200:210,200:210] = [255,255,255]

plt.imshow(arr, origin='lower')

plt.title("Milky Way Illistration with Point Source")

plt.savefig("galaxy_with_point.png")
plt.show()
```

```python
# arr = np.sum(arr, axis=2)

src = Source(array=arr)
src.pad(1000)
```

```python
df = EinsteinDeflector(theta_E=1000)
deflected = df.get_image(src)
```

```python
plt.imshow(deflected/np.max(deflected), origin='lower')

plt.title("Lensed Milky Way Illistration with Point Source")

plt.savefig("galaxy_distortion_with_point.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from power_law import PowerLawDeflector
from base_classes import Source, Deflector, DiskSource
```

```python
sc = DiskSource(1001, 50)
plane = sc.array
cbar = plt.imshow(plane, origin='lower')
plt.colorbar(cbar)

plt.title("uniform disk source")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

```python
plt.figure(figsize=(6,6))
src = DiskSource(100, 50)
df = PowerLawDeflector(50,2)

kappa = df.get_convergence(src)

# cbar = plt.imshow(kappa, origin='lower')
# plt.colorbar(cbar)
# plt.show()

plt.contour(kappa)
plt.show()
```

```python
df = PowerLawDeflector(400, 2)
image = df.get_image(source=sc)

cbar = plt.imshow(image, origin='lower')
plt.colorbar(cbar)

plt.title("deflection of a uniform disk about an isothermal, spherical lens")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from piemd import PIEMD
from base_classes import Source, DiskSource
import scipy as sp
from PIL import Image
```

Lets take an image of a galaxy and make a source and lens

```python
img = np.array(Image.open("figures_in/hubble_galaxy1.jpeg"))
# greyscale
img = np.sum(img, axis=-1)
# make it square
img = img[:, 17:221]
print(img.shape)

cbar = plt.imshow(img, origin='lower', cmap='inferno')
plt.colorbar(cbar)
plt.show()
```

```python
src = Source(img)
src.pad(3000)
# src.plot_source()
```

```python
df = PIEMD(src, theta_E=2000, q=0.85, s=50, phi=np.pi * 3/4)
# lens = df.generate_lens()
# plt.imshow(lens)
# plt.show()
```

```python
img_hubble = df.generate_hubble_image(sigma=2000/3)
```

```python
cbar = plt.imshow(img_hubble, cmap='inferno', origin='lower')
plt.colorbar(cbar)
plt.savefig("figures_out/hubble_image1.png")
plt.show()
```

```python
img = np.array(Image.open("figures_in/Milky_Way_illustration-1.jpeg"))
# greyscale
img = np.sum(img, axis=-1)
# make it square
print(img.shape)
img = sp.ndimage.zoom(img, 1/3)
print(img.shape)
```

```python
src = Source(img)
src.pad(3000)

df = PIEMD(src, theta_E=2000, q=0.6, s=400, phi=np.pi * 1/4)
df.align_source(cx=3689, cy=3689)
```

```python
img = df.generate_image()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
plt.imshow(img, origin='lower', cmap='inferno')
plt.show()
```

```python
img_hubble = df.generate_hubble_image(sigma=2000/3)
```

```python
cbar = plt.imshow(img_hubble, cmap='inferno', origin='lower')
plt.colorbar(cbar)
plt.savefig("figures_out/hubble_image2.png")
plt.show()
```

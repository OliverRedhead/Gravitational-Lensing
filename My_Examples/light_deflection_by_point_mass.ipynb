````python
from scipy import special as sy 
import numpy as np
import matplotlib.pyplot as plt

from point_mass import PointMass

from astropy import constants as const
````

Consider the following example simulating the deflection angle around the sun. We expect about 1.75''. We can see our approximations give a good result yay.

````python
M = const.M_sun.value       # type: ignore
pm = PointMass(M)

xi = np.linspace(1, 20, 1000)*const.R_sun.value       # type: ignore
alpha = pm.get_deflection(xi, arcseconds=True)

plt.plot(xi,alpha)
plt.xlabel(r"impact parameter $(\xi)$ [m]")
plt.ylabel(r"deflection $(\alpha)$ [arcseconds]")

plt.title("Deflection angle of point mass")
plt.show()
````

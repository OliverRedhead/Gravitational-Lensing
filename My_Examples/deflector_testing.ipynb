```python
from deflector import Deflector
import matplotlib.pyplot as plt
import numpy as np
```

```python
file = '../Meneghetti_Examples/mass_distributions/kappa.01.023.fits'
```

```python
df = Deflector(file,pad=True, padwidth=1)
```

```python
plt.imshow(df.image(), origin='lower')
plt.show()
```

```python
ax,ay = df.deflection_map()

cbar=plt.imshow(ax, origin='lower', cmap='plasma')
plt.colorbar(cbar)
plt.show()

cbar=plt.imshow(ay, origin='lower', cmap='plasma')
plt.colorbar(cbar)
plt.show()
```

```python
cbar=plt.imshow(np.sqrt(ax**2 + ay**2), origin='lower', cmap='plasma')
plt.colorbar(cbar)
plt.show()
```

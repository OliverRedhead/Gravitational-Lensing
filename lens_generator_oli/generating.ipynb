```python
from dspl_generator import DSPLGenerator

import matplotlib.pyplot as plt
```

```python
amp=1.3

gen = DSPLGenerator(
    fov=20,
    lens_img="data/euclid_deep_VIS2.npy",
    lens_amp=1.0*amp,
    source1_img=None,
    source1_amp=3.0*amp,
    source2_img=None,
    source2_amp=3.0*amp,
    telescope="euclid",
    exposure_time=600
)
```

```python
gen.shuffle()
```

```python
sim = gen.get_simulation()
plt.imshow(sim, cmap="PuOr_r", extent=gen.pixel_grid.extent)
plt.colorbar()
plt.show()
```

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

import scipy as sp
from PIL import Image
```

```python
img = np.array(Image.open("hubble_deep.jpg"))
# img = np.sum(img, axis=-1)
```

```python
%matplotlib inline
import matplotlib.pyplot
```

```python
d=400

x=743
y=509

plt.imshow(img[y-d:y+d, x-d:x+d], origin='lower')
plt.savefig("spiral_1.png")
np.save("spiral_1.npy",np.sum(img[y-d:y+d, x-d:x+d], axis=-1) )
plt.show()
```

```python
spiral = np.sum(img[y-d:y+d, x-d:x+d], axis=-1)
plt.imshow(spiral, origin='lower')
plt.colorbar()
plt.show()

sigma = np.std(spiral)

no_noise = spiral.copy()
no_noise[no_noise < 2*sigma] = 0
plt.imshow(no_noise, origin='lower')
plt.colorbar()
plt.show()


noise = spiral - no_noise
plt.imshow(noise, origin='lower')
plt.colorbar()
plt.show()
```

```python
no_noise[700:,700:] = no_noise[700:,700:] / 1.5 

plt.imshow(no_noise, origin='lower')
plt.show()

np.save("spiral_1.npy", no_noise)
```

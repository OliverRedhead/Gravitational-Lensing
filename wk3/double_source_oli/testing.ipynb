```python
from double_source import DoubleSource
from base_classes import Source, Plane, Deflector, SIS, PIEMD, DiskSource

import numpy as np
import matplotlib.pyplot as plt
```

```python
size = 300

s1 = DiskSource(size, 10)
s2 = DiskSource(size, 5)

p1 = Plane(source=s1)
p2 = Plane(source=s2)

theta_E1 = 40
theta_E2 = 70

df = PIEMD(size, theta_E=None, q=0.6, phi=np.pi/4, s=2)
```

```python
dbl_source = DoubleSource(df, p1, p2, theta_E1, theta_E2)
```

```python
img = dbl_source.get_image()
```

```python
plt.imshow(img, origin='lower')
plt.colorbar()
plt.savefig("figures_out/double_ring.png")
plt.show()
```
